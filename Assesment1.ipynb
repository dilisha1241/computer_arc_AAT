```python
def to_binary(num, bits):
    """Convert a signed integer to 2's complement binary."""
    return format(num & ((1 << bits) - 1), f'0{bits}b')


def signed_arithmetic(a, b, bits=16):
    """Perform signed subtraction using 2's complement."""
    mask = (1 << bits) - 1

    # Represent operands in 2's complement
    A = a & mask
    B = b & mask

    # Subtraction: A - B = A + 2's complement of B
    B_complement = ((~B) + 1) & mask
    result = (A + B_complement) & mask

    # Convert result to signed decimal
    if result & (1 << (bits - 1)):
        decimal_result = result - (1 << bits)
    else:
        decimal_result = result

    # Check signed arithmetic overflow
    overflow = ((A ^ B) & (A ^ result) & (1 << (bits - 1))) != 0

    print("Operand A (8-bit):", to_binary(a, 8))
    print("Operand B (8-bit):", to_binary(b, 8))
    print("Operand A (16-bit):", to_binary(a, 16))
    print("Operand B (16-bit):", to_binary(b, 16))
    print("2's complement of B:", to_binary(B_complement, 16))
    print("Result (16-bit):", to_binary(result, 16))
    print("Result (decimal):", decimal_result)
    print("Arithmetic Overflow:", "Yes" if overflow else "No")
    print("-" * 45)


print("a) 35 - (+29)")
signed_arithmetic(35, 29)

print("b) (-35) - (-12)")
signed_arithmetic(-35, -12)
```

```python
def rotate_left_without_carry(binary, positions):
    print("\nRotate Left without Carry (ROL)")
    print("Original binary number:", binary)

    for i in range(positions):
        msb = binary[0]
        binary = binary[1:] + msb
        print(f"After rotation {i + 1}: {binary}")

    print("Final result:", binary)


def rotate_left_with_carry(binary, carry, positions):
    print("\nRotate Left with Carry (RCL)")
    print("Original binary number:", binary)
    print("Initial carry flag:", carry)

    for i in range(positions):
        msb = binary[0]
        binary = binary[1:] + carry
        carry = msb

        print(f"After rotation {i + 1}: {binary}, Carry = {carry}")

    print("Final result:", binary)
    print("Final carry flag:", carry)


# a) ROL by 2 positions
rotate_left_without_carry("11110110", 2)

# b) RCL by 2 positions, initial carry = 0
rotate_left_with_carry("10000110", "0", 2)
```
